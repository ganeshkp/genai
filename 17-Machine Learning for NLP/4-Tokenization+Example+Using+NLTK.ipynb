```python
corpus="""Hello, I am a student learning Python.
I enjoy solving problems and building projects.
Python is a versatile language that is great for beginners.
I also like to explore data science and machine learning.
I often use libraries like NumPy, Pandas, and Matplotlib.
I believe that practice is key to mastering programming.
"""
```

```python
print(corpus)
```

```python
##  Tokenization
## Sentence-->paragraphs
from nltk.tokenize import sent_tokenize
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
documents=sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
## Tokenization 
## Paragraph-->words
## sentence--->words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer=TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```
