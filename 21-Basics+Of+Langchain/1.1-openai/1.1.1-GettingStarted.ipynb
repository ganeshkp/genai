```python
### Chatprompt Template
from langchain_core.prompts import ChatPromptTemplate

prompt=ChatPromptTemplate.from_messages(
    [
        ("system","You are an expert AI Engineer. Provide me answers based on the questions"),
        ("user","{input}")
    ]

)
prompt
```

```python
## chain
chain=prompt|llm

response=chain.invoke({"input":"Can you tell me about Langsmith?"})
print(response)
```

```python
## stroutput Parser

from langchain_core.output_parsers import StrOutputParser
output_parser=StrOutputParser()
chain=prompt|llm|output_parser

response=chain.invoke({"input":"Can you tell me about Langsmith?"})
print(response)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableLambda
from langchain_openai import ChatOpenAI  # or your LLM

# Create the LLM
# llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0)
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)  # Use the model you prefer

# Create the prompt
prompt=ChatPromptTemplate.from_messages(
    [
        ("system","You are an expert AI Engineer. Provide me answers based on the questions"),
        ("user","{input}")
    ]
)

# Extract only the text field from LLMChain output
extract_text = RunnableLambda(lambda x: x.content if hasattr(x, "content") else x)

# StrOutputParser (not strictly necessary here, but safe to keep)
output_parser = StrOutputParser()

# Full pipeline
chain = prompt | llm | extract_text | output_parser

# Run it
response = chain.invoke({"input": "Can you tell me about Langsmith?"})
print(response)  # This will be plain text, no extra dict
```
